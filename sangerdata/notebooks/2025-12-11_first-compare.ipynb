```python
import os
import pandas as pd
```

```python
df_samples = pd.read_csv("../sample_set/table.zambia_sanger.all_info.csv", dtype={"sample_id":str})
```

```python
df_samples
```

```python
SOURCE_DIR = "../results_newprotocol/4csvs/"

dfs = []
for csv in os.listdir(SOURCE_DIR):
    try:
        df = pd.read_csv(f"{SOURCE_DIR}/{csv}", dtype={"sample_id": str})
    except:
        print(f"Failed for {csv}!")
        continue

    dfs.append(df)
df_concat = pd.concat(dfs)
```

```python
(724 - 42) * 3
```

```python
df_concat["sanger_A724E"] = [
    "724A>724E" in csq
    for csq in df_concat["csq"]
]
```

```python
df_concat
```

```python
df_merged = pd.merge(
    left=df_samples,
    right=df_concat.groupby("sample_id").sanger_A724E.sum().reset_index(),
)
```

```python
pd.crosstab(
    df_merged["sanger_A724E"],
    df_merged["gt_A724E"],
    dropna=False
)
```
