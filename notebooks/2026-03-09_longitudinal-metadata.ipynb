```python
import pandas as pd
```

```python
metadata_mis2018_df = pd.read_csv("../metadata/2026-03-09_longitudinal-kaoma-bb/mis2018_metadata_luampa_nkeyema_kaoma.csv")
```

```python
metadata_mis2018_df["district"].value_counts()
```

```python
metadata_mis2018_df["district"] = metadata_mis2018_df["district"].str.capitalize()
```

```python
metadata_mis2018_kaoma_df = metadata_mis2018_df[metadata_mis2018_df["district"] == "Kaoma"]
```

```python
metadata_mis2018_kaoma_df = metadata_mis2018_kaoma_df[["LabID", "province", "district"]]
metadata_mis2018_kaoma_df["year"] = 2018
metadata_mis2018_kaoma_df.rename(columns={"LabID": "sample_id"}, inplace=True)
```

```python
metadata_tes2021_df = pd.read_csv("../seqdata/all_standard/results/2026-01-12_MM099_TES2021KAOMA_Batch1/metadata/samples.csv")
```

```python
metadata_tes2021_kaoma_df = metadata_tes2021_df[metadata_tes2021_df["sample_type"] == "Field"]
```

```python
metadata_tes2021_kaoma_df = metadata_tes2021_kaoma_df[["sample_id"]]
metadata_tes2021_kaoma_df["province"] = "Western"
metadata_tes2021_kaoma_df["district"] = "Kaoma"
metadata_tes2021_kaoma_df["year"] = 2021
```

```python
metadata_tes2023_df = pd.read_csv("../seqdata/all_standard/results/2026-01-18_MM100_TES2023KAOMA_Batch1/metadata/samples.csv")
```

```python
metadata_tes2023_kaoma_df = metadata_tes2023_df[metadata_tes2023_df["sample_type"] == "Field"]
```

```python
metadata_tes2023_kaoma_df = metadata_tes2023_kaoma_df[["sample_id"]]
metadata_tes2023_kaoma_df["province"] = "Western"
metadata_tes2023_kaoma_df["district"] = "Kaoma"
metadata_tes2023_kaoma_df["year"] = 2023
```

```python
metadata_mis2024_hrp23_df = pd.read_csv("../metadata/2026-01-23_consolidated-metadata/table.zambia_all_data_combined.csv")
```

```python
metadata_mis2024_hrp23_df["district"].value_counts()
```

```python
metadata_mis2024_hrp23_kaoma_df = metadata_mis2024_hrp23_df[metadata_mis2024_hrp23_df["district"] == "kaoma"]
```

```python
metadata_mis2024_hrp23_kaoma_df = metadata_mis2024_hrp23_kaoma_df[["sample_id", "province", "district"]]
metadata_mis2024_hrp23_kaoma_df["year"] = 2024
```

```python
ex_vivo_paths = [
    "../seqdata/all_standard/results/2026-04-07_BM013_ex_vivo_Batch1/metadata/samples.csv",
    "../seqdata/all_standard/results/2026-04-21_BM014_ex_vivo_Batch2/metadata/samples.csv",
    "../seqdata/all_standard/results/2026-04-27_BM015_ex_vivo_Batch3/metadata/samples.csv",
    "../seqdata/all_standard/results/2026-04-27_BM016_ex_vivo_Batch4/metadata/samples.csv",
]
```

```python
metadata_2026_ex_vivo_dfs = [pd.read_csv(path, dtype={"sample_id": str}) for path in ex_vivo_paths]
```

```python
metadata_2026_ex_vivo_df = pd.concat([df[df["sample_type"] == "Field"] for df in metadata_2026_ex_vivo_dfs], ignore_index=True)
```

```python
metadata_2026_ex_vivo_df.duplicated(subset=["sample_id"]).sum()
```

```python
metadata_2026_ex_vivo_df[metadata_2026_ex_vivo_df.duplicated(subset=["sample_id"], keep=False)]
```

```python
metadata_2026_ex_vivo_df
```

```python
metadata_2026_ex_vivo_df.sample_id.str.contains("-D3").sum()
```

```python
metadata_2026_ex_vivo_df = metadata_2026_ex_vivo_df[~metadata_2026_ex_vivo_df["sample_id"].str.contains("-D3")]
metadata_2026_ex_vivo_df.sample_id.str.contains("-D3").sum()
```

```python
metadata_2026_ex_vivo_df.duplicated(subset=["sample_id"]).sum()
```

```python
metadata_2026_ex_vivo_df = metadata_2026_ex_vivo_df[["sample_id"]]
metadata_2026_ex_vivo_df["province"] = "Western"
metadata_2026_ex_vivo_df["district"] = "Kaoma"
metadata_2026_ex_vivo_df["year"] = 2026
```

```python
metadata_kaoma_all_df = pd.concat([
    metadata_mis2018_kaoma_df,
    metadata_tes2021_kaoma_df,
    metadata_tes2023_kaoma_df,
    metadata_mis2024_hrp23_kaoma_df,
    metadata_2026_ex_vivo_df,
], ignore_index=True)
```

```python
metadata_kaoma_all_df.to_csv("../metadata/2026-03-09_longitudinal-kaoma-bb/metadata_kaoma_all.csv", index=False)
```
